## Research question

How can query coverage and retrieval provenance be made inspectable across terpene and topic searches in PubMed?

This notebook is an auditable starting point. It reports only what the local, versioned input artifacts can support; it does not create missing experimental evidence.

```python
from pathlib import Path
import csv, json

ROOT = Path("..")
manifest = ROOT / "data" / "manifest.tsv"
print("Manifest present:", manifest.exists())
if manifest.exists():
    with manifest.open(newline="") as f:
        rows = list(csv.DictReader(f, delimiter="\t"))
    print("Sources:", len(rows))
    for row in rows:
        print(row.get("source_id"), row.get("accession_or_url"), row.get("local_path"))
```
